```python
import pandas as pd

# Load the data using the correct nested paths
train_df = pd.read_csv('/kaggle/input/competitions/house-prices-advanced-regression-techniques/train.csv')
test_df = pd.read_csv('/kaggle/input/competitions/house-prices-advanced-regression-techniques/test.csv')

# Select the features
features = ['LotArea', '1stFlrSF', 'TotRmsAbvGrd']
X_train = train_df[features].values
y_train = train_df['SalePrice'].values
X_test = test_df[features].values

print("Data loaded successfully!")
```

```python
import numpy as np

def z_score_normalize(X):
    """
    Computes the z-score normalisations for feature matrix x.
    """
    mean=np.mean(X,axis=0)
    std=np.std(X,axis=0)
    X_norm=(X-mean)/std
    return X_norm

X_train_norm=z_score_normalize(X_train)
X_test_norm=z_score_normalize(X_test)
print("Normalized Training Features Sample:\n", X_train_norm[:5])
print("\nNew column means (should be close to 0):", np.mean(X_train_norm, axis=0))
print("New column standard deviations (should be 1):", np.std(X_train_norm, axis=0))
```

```python
def compute_cost(X, y, w, b):
    """
    Computes the cost function for linear regression.
    X: (m, n) normalized feature matrix
    y: (m,) target values (SalePrice)
    w: (n,) model weights
    b: scalar model bias
    """
    m = X.shape[0]  # Fixed: changed lowercase 'x' to uppercase 'X'
    
    # Step 1: Calculate predictions f_wb = Xw + b
    predictions = np.dot(X, w) + b
    
    # Step 2: Calculate squared errors
    squared_error = (predictions - y) ** 2
    
    # Step 3: Compute total cost J(w,b)
    total_cost = (1 / (2 * m)) * np.sum(squared_error)
    return total_cost

def compute_gradient(X, y, w, b):
    """
    Computes the gradient for linear regression.
    """
    m, n = X.shape
    
    # Calculate error vector: (f_wb - y)
    error = (np.dot(X, w) + b) - y
    
    # Vectorized gradient calculations
    dj_dw = (1 / m) * np.dot(X.T, error)
    dj_db = (1 / m) * np.sum(error)
    
    return dj_dw, dj_db
```

```python
# Initialize parameters
m, n = X_train_norm.shape
w_init = np.zeros((n,))
b_init = 0.0

# Set hyperparameters
iterations = 1000
alpha = 0.1  # Learning rate works well now thanks to Z-score scaling!

# Copy initial parameters
w = np.copy(w_init)
b = b_init

print(f"Starting Cost: {compute_cost(X_train_norm, y_train, w, b):,.2f}")

# Gradient Descent Loop
for i in range(iterations):
    # Calculate gradients (Fixed: correctly unpacking both dj_dw and dj_db)
    dj_dw, dj_db = compute_gradient(X_train_norm, y_train, w, b)
    
    # Update parameters simultaneously
    w = w - alpha * dj_dw
    b = b - alpha * dj_db
    
    # Print cost progress every 100 iterations
    if i % 100 == 0:
        cost = compute_cost(X_train_norm, y_train, w, b)
        print(f"Iteration {i:4d}: Cost = {cost:,.2f}")

print("\nTraining complete!")
print("Final Optimized Weights (w):", w)
print("Final Optimized Bias (b):", b)

```
